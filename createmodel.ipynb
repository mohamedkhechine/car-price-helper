```python
df_tun = df_tun.drop(columns=['horsepower'])
```

```python
common_cols = [
    col for col in df_tun.columns
    if col in df_turkey.columns and col != 'price'
]

x_tun_common = df_tun[common_cols].to_numpy()
x_turk_common = df_turkey[common_cols].to_numpy()

tunisia_only_cols = [
    col for col in df_tun.columns
    if col not in df_turkey.columns and col != 'price'
]

x_tun_extra = df_tun[tunisia_only_cols].to_numpy()

y_tun = df_tun['price'].to_numpy()
y_turk = df_turkey['price'].to_numpy()
```

```python
x_turk_common = x_turk_common.astype('float32')
y_turk = y_turk.astype('float32')
```

```python
from tensorflow.keras import Model, Input
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.models import Sequential

commonbranch = Sequential([
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(16, activation='relu')
])

extrabranch = Sequential([
    Dense(32, activation='relu'),
    Dense(16, activation='relu')
])

finalbranch = Sequential([
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(1, activation='linear')
])

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=20,
    restore_best_weights=True
)

# Create an explicit input
common_input = Input(shape=(x_turk_common.shape[1],))

# Pass it through the pretrained/common branch
common_features = commonbranch(common_input)

# Temporary output for Turkey price prediction
turkey_output = Dense(1, activation='linear')(common_features)

# Build the Turkey pretraining model
turkeypremodel = Model(
    inputs=common_input,
    outputs=turkey_output
)

turkeypremodel.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

history = turkeypremodel.fit(
    x_turk_common,
    y_turk,
    validation_split=0.2,
    epochs=1000,
    batch_size=32,
    callbacks=[early_stop]
)

# Save ONLY the common branch weights
commonbranch.save_weights("common_weights.weights.h5")
```

```python
print(x_turk_common.dtype)
print(y_turk.dtype)

print(x_turk_common.shape)
print(y_turk.shape)
```

```python

from tensorflow.keras.layers import Concatenate

# Freeze the Turkey-trained common branch
commonbranch.trainable = False

# Inputs
tun_common_input = Input(shape=(x_tun_common.shape[1],))
tun_extra_input = Input(shape=(x_tun_extra.shape[1],))

# Common branch: uses Turkey-trained weights
common_features = commonbranch(tun_common_input)

# Extra branch: starts with fresh weights
extra_features = extrabranch(tun_extra_input)

# Combine both representations
combined = Concatenate()([common_features, extra_features])

# Final branch predicts Tunisian price
output = finalbranch(combined)

# Complete Tunisia model
tunisia_model = Model(
    inputs=[tun_common_input, tun_extra_input],
    outputs=output
)

tunisia_model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

early_stop_tun = EarlyStopping(
    monitor='val_loss',
    patience=20,
    restore_best_weights=True
)

history_tun = tunisia_model.fit(
    [x_tun_common, x_tun_extra],
    y_tun,
    validation_split=0.2,
    epochs=1000,
    batch_size=32,
    callbacks=[early_stop_tun]
)
```
