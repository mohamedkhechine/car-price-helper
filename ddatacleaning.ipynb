```python
print(cars['horsepower'].unique())
```

```python
cleaned=[]
for row in cars['horsepower']:
    if row=="plus de 48 cv":
        row=row.replace("plus de","")
    row=row.replace("cv","")
    row=row.strip()
    cleaned.append(int(row))
cars['horsepower']=cleaned
cars.head()
```
